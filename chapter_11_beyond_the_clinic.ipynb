# Chapter 11 — Beyond the Clinic

The loop does not change when it leaves the ward. The *social object* does. A P300 speller that Amina commands is an assistive device. The same family of sensors, sold as a focus score to an employer, is a workplace monitor. This chapter keeps the engineering honest and names the extra person who has entered Stage S8.

**Learning outcomes**

1. Place consumer EEG, games, XR, and wellness products on the resolution–risk surface of Chapter 3.
2. Distinguish a medical claim from a wellness claim without using the vendor’s vocabulary.
3. Treat passive BCI as state estimation wrapped in a policy that may serve someone other than the wearer.
4. Write a consent and data-flow paragraph a workplace system would need before it could appear in this book as finished.
5. Refuse enhancement slogans that violate the lexicon of Chapter 8.


---

## 11.1 Same pipeline, different contract

Stages S0–S9 still apply. What changes is:

- who chooses that the loop exists;  
- who sees $\hat{y}$;  
- what $\hat{y}$ is allowed to do;  
- what happens when the wearer says no.

> **Definition 11.1 (Wearer-directed loop).**  
> The person whose nervous system is measured can start, stop, and inspect the output, and is the principal beneficiary.

> **Definition 11.2 (Other-directed loop).**  
> A second party — employer, school, platform, parent, insurer — receives or acts on $\hat{y}$.

Amina’s speller is 11.1. A dashboard of “engagement” on a manager’s screen is 11.2. Many products advertise 11.1 and implement 11.2 in the privacy policy.

Cognitive computing appears here almost only as **policy over other people**: adaptive game difficulty, attention dashboards, hiring screens. That is still Definition 8.1. It is not a reason to relax Definitions 7.3 or 2.3.


---

## 11.2 Where consumer hardware sits

Dry EEG headbands occupy the low-information, low-medical-risk corner of Chapter 3. They add new risks that are not surgical:

- artefact sold as intent (Chapter 6);  
- state sold as content (Definition 8.4);  
- data leaving the room;  
- function creep from wellness to evaluation.

MEG and implants do not belong in this chapter as consumer objects. If a firm implies they do, it is using Chapter 3’s vocabulary as costume.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7.4, 4.4))
ax.set_xlim(0, 6)
ax.set_ylim(0, 6)
ax.set_xlabel("other-directed power over the wearer")
ax.set_ylabel("strength of the medical or assistive claim")
ax.set_title("Two axes consumer copy usually collapses")
items = {
    "Amina's home speller": (0.7, 4.6),
    "Personal meditation log": (0.8, 1.3),
    "Adaptive game (local only)": (1.2, 1.0),
    "Classroom focus traffic-light": (3.8, 1.6),
    "Employer attention score": (5.0, 1.8),
    "Depression-treatment headband": (2.0, 4.2),
}
for name, (x, y) in items.items():
    ax.scatter([x], [y], s=60, color="#4c6a92")
    ax.text(x + 0.08, y + 0.08, name, fontsize=8)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

A product can be low on the vertical axis and still be unacceptable far to the right. Do not let a weak scientific claim launder a strong social one.

## 11.3 Medical claim versus wellness claim

Regulators draw this line differently across jurisdictions. The book needs a working test, not a lawyer’s treatise.

> **Definition 11.3 (Medical claim, working test).**  
> The product is offered to diagnose, treat, prevent, or manage a disease or a clinical deficit, or to replace a function in the sense of Chapter 9.

> **Definition 11.4 (Wellness claim, working test).**  
> The product is offered to a healthy wearer for relaxation, entertainment, or unspecific “optimisation,” without a disease target.

Consequences:

- 11.3 demands clinical evidence of the Chapter 10 kind.  
- 11.4 still demands that advertised *mechanisms* not be false. “Reads your thoughts to reduce stress” fails Chapter 8 even as wellness.  
- Moving a slider from 11.4 to 11.3 in marketing without moving the evidence is the usual failure.

Students will meet both in internships. The design sheet must have a box: *claim class*.

## 11.4 Games, XR, and passive adaptation

Games and extended reality are legitimate testbeds for reactive and active paradigms. They also hide gaze and EMG unusually well because the user is already moving.

Passive adaptation — difficulty, music, scene lighting driven by an inferred state — is Definition 1.4 plus a policy. Allowed questions:

- Does $\hat{y}$ beat a nuisance decoder (Chapter 6)?  
- Does the policy make the game better *for the wearer*?  
- Is $\hat{y}$ stored, and who can buy it later?

If the platform retains neural features to train a foundation model (Chapter 7), the wearer has entered a data labour relationship. Write that down.

## 11.5 Workplaces and classrooms

Here Definition 11.2 is the default, however soft the branding.

Problems that are not solved by a better classifier:

- **Consent under hierarchy.** An employee can sign and not be free.  
- **Construct invalidity.** “Focus” is not a single neural generator.  
- **Punishment by proxy.** A low score becomes a shift assignment.  
- **Leakage of extras.** Affect, sleep, pregnancy-related change, seizure risk — variables nobody named in the job description.

> **Definition 11.5 (Neural evaluation).**  
> Use of a decoded neural variable to judge, rank, or gate a person’s access to work, school, insurance, or a platform.

This book treats 11.5 as unfinished as a product category. Not because the ethics chapter has not been reached, but because the *measurement* does not support the *decision* (Definitions 2.3 and 8.4).

If you build one anyway, the minimum paragraph is:

1. wearer can refuse without professional penalty;  
2. $\hat{y}$ stays on-device unless a named purpose is accepted;  
3. no secondary sale;  
4. a human decision-maker is identified;  
5. a deletion date.


---

## 11.6 Enhancement

Enhancement — better memory, faster decisions, extra channels of perception — is a purpose theme from Chapter 1 and a marketing theme from everywhere else.

For this introductory text:

- non-invasive consumer devices do not currently deliver content-level enhancement (Chapter 8);  
- closed-loop stimulation that might one day do more is a Chapter 3 writing problem with Chapter 12 consent;  
- fairness: an enhancement that only a surgical elite can buy is a social design, not only a device design.

Do not use “unlock human potential” as a methods sentence.


---

## 11.7 Worked placements

**Meditation headband, log stays on the phone.** 11.1, 11.4, low vertical claim. Still forbid “reads thoughts.”

**Esports EEG used only to change game difficulty locally.** 11.1 plus passive policy. Artefact audit required.

**Call-centre “empathy and focus” score visible to a supervisor.** 11.2 and 11.5. Not a finished design in this book.

**Insurance discount for nightly “brain health” uploads.** 11.2 with a financial penalty for refusal. Name the coercion.

**XR plus implant for a paralysed user to control a home OS.** That is Chapter 9 wearing a headset. Do not file it under wellness.

Chapter 12 will generalise consent, data, abandonment, and translation. Carry concrete objects.

1. Wearer-directed versus other-directed.  
2. Medical versus wellness as a claim test.  
3. Neural evaluation as a decision the measurement cannot bear.  
4. Data labour when features train someone else’s model.

## 11.9 Chapter summary

Beyond the clinic the loop is cheap to mount and expensive to govern. Consumer sensors sit at low information grain and manufacture new risks of artefact, over-claim, and secondary use. Wellness language does not license false mechanisms. Passive adaptation can serve the wearer or a supervisor; those are different loops. Workplace and classroom scores that judge people are, in this book, an unfinished category because state is not content and consent under hierarchy is not consent. Enhancement slogans are refused until a usable trace and a fair allocation are both on the design sheet.
