## 3.1 The design surface

Every interface occupies a point in a space whose axes include:

- spatial grain (how local is the measurement or the stimulus?);
- temporal grain (how fast can the variable change and still be seen?);
- information potential (how many independent control dimensions are plausible?);
- risk (infection, haemorrhage, seizure, explantation, abandoned hardware);
- logistics (setup minutes, technician, shielding, hospital, home);
- stability across days.

These axes are coupled. Moving toward finer spatial grain usually moves toward higher risk and heavier logistics. The coupling is not a law of nature in every detail, but it is reliable enough to govern first-semester design.

> **Definition 3.1 (Resolution–risk surface).**  
> The set of feasible recording and stimulation choices, plotted by the information they can plausibly support against the burden they impose on a named user.

Amina at home with a caregiver is not the same point as a healthy student under a wet cap in a laboratory, and neither is the same point as a trial participant with an intracortical array. Chapter 1 asked you to name the user first. This chapter is why.

## 3.2 Recording: what is actually measured

### Electroencephalography (EEG)

Metal electrodes on the scalp record voltage differences produced by volume-conducted fields (Chapter 2). Temporal grain is excellent (milliseconds). Spatial grain is poor: each channel is a centimetre-scale, skull-filtered mixture, biased toward aligned superficial sources.

**Strengths.** Non-invasive; affordable; usable outside a scanner; the workhorse of P300, SSVEP, and motor-imagery teaching laboratories.

**Limits.** Low signal-to-noise ratio; muscular and ocular artefacts that can mimic “thought control”; long setup for research-grade wet systems; dry consumer headbands trade signal quality for convenience.

**Honest use.** Communication when other channels fail; rehabilitation research; passive state estimation at coarse grain. Not a transcript of working-memory contents.

### Magnetoencephalography (MEG)

SQUID or optically pumped sensors record magnetic fields associated with the same currents. Spatial grain can be finer than EEG for some source configurations. Temporal grain remains millisecond-scale.

**Limits.** Shielding, cost, immobility. MEG is a research instrument, not a home communication device.

### Functional near-infrared spectroscopy (fNIRS)

Near-infrared light estimates haemodynamic change in superficial cortex. Spatial grain is intermediate and shallow. Temporal grain is slow (seconds), because blood follows neurons at a delay.

**Honest use.** Complementary to EEG in hybrid systems; coarse workload or localiser studies. Not a high-rate command channel.

### Electrocorticography (ECoG)

Electrodes on the cortical surface, beneath the skull. Spatial grain is finer than EEG; artefacts from muscle and blink are reduced; bandwidth extends higher.

**Cost.** Craniotomy or related surgery. Often available first in patients who already require implantation for epilepsy monitoring. That convenience is not a sampling of “the brain”; it is a sampling of those cortices a surgeon can reach in those patients.

### Intracortical microelectrodes

Penetrating arrays record spikes of one or a few nearby cells and the local field potential. Information potential for kinematics and attempted speech is the highest currently demonstrated in humans.

**Cost.** Open or minimally open neurosurgery; infection and haemorrhage risk; chronic tissue response (scarring, micromotion, falling unit yield); obligation to support the person after the paper is published.

### Endovascular sensors

A device delivered through blood vessels to sit near cortex records fields from a vascular vantage. The aim is to approach ECoG-like information with a different surgical path.

**Honest status.** A distinct point on the surface — not “non-invasive,” not automatically equivalent to a penetrating array. Evaluate it as itself.

### Other measurements

Functional MRI has excellent spatial maps and poor temporal grain and no home use. Invasive depth electrodes and deep-brain leads can record from subcortical structures in clinical contexts. They enlarge the atlas of Chapter 2; they do not cancel the surface.


### Comparison at a glance

| Modality | Quantity | Spatial grain | Temporal grain | Typical setting | Dominant risk |
|---|---|---|---|---|---|
| Consumer EEG | Scalp field | Very coarse | ms | Home, office | Over-claim; artefact as “intent” |
| Research EEG | Scalp field | Coarse | ms | Lab, clinic | Low medical; high false-decode |
| fNIRS | Haemodynamics | Shallow, coarse | seconds | Lab | Low medical; slow loop |
| MEG | Magnetic field | Finer than EEG | ms | Scanner suite | Cost, immobility |
| ECoG | Cortical-surface field | Intermediate | ms | Theatre, epilepsy ward | Surgery |
| Endovascular | Field near vessel | Intermediate | ms | Interventional suite | Vascular procedure |
| Microelectrode array | Spikes + LFP | Fine | sub-ms to ms | Neurosurgery | Surgery + chronic tissue |

> **Definition 3.2 (Channel).**  
> In this book a *channel* is one recorded time series after analogue front-end and digitisation: one EEG electrode versus reference, one ECoG contact, one microelectrode, one fNIRS optode pair. “64 channels” is a hardware count, not a count of independent brain sources.

## 3.3 The analogue front-end, before algorithms begin

A student who jumps to classifiers skips the reason most demonstrations fail in a cafeteria.

1. **Contact.** Electrolyte gel, dry polymer, or implanted metal. Impedance and motion decide whether the trace is neural or mechanical.
2. **Amplification and filtering.** Fields at the scalp are tens of microvolts. Line noise at 50 or 60 Hz is not optional in most buildings.
3. **Digitisation.** Sampling rate must respect the band you claim to use. Claiming gamma features at 128 Hz is a contradiction.
4. **Referencing.** EEG voltages are differences. An unlucky reference imports artefact into every channel.
5. **Grounding and isolation.** Safety standards exist because stimulators and mains electricity are not theoretical.

Chapter 6 will treat digital preprocessing. Nothing in Chapter 6 can recover information destroyed here.

## 3.4 Stimulation: writing is not reading backwards

> **Definition 3.3 (Neural stimulation).**  
> The delivery of energy — most often electrical — intended to change neuronal activity.

Reading estimates a state from a field or a spike train. Writing imposes current that spreads through tissue according to electrode geometry and conductivity. The volume you stimulate is not the volume you recorded from, even on the same shank. Adaptation, phosphenes, afterdischarges, and mood change are not “decode error.” They are properties of stimulation.

**Cousins that are not automatically BCIs.**  
Cochlear implants and many deep-brain stimulators are neural interfaces and neuroprostheses. They become part of a BCI story when measured neural activity *governs* the stimulation, or when stimulation closes a loop with a decoder (Chapter 1, direction of information). A fixed-schedule stimulator is therapy. It is not thereby an interface.

**Forms a first-semester student should be able to name.**

- **Cortical surface stimulation** — used in mapping and in some bidirectional research; can evoke somatosensory or visual fragments.
- **Intracortical microstimulation** — finer grain; still not a pixel-perfect tactile display.
- **Deep-brain stimulation** — established in movement disorders and expanding in psychiatry; typically subcortical; closed-loop variants exist.
- **Non-invasive stimulation** (TMS, transcranial electrical stimulation) — useful in research and in some therapy; coarse; sham control is mandatory in claims of enhancement.

Cognitive computing can supply a *policy* for when to stimulate (for example, detect a state and intervene). It cannot make a coarse current into a precise memory edit.

## 3.5 Bidirectional systems

A bidirectional interface records and stimulates in one loop. The scientific attraction is obvious: restore touch to a robotic hand; interrupt a seizure; pair imagined movement with a pulse that the cortex can learn from.

The engineering difficulties are also obvious:

- stimulation artefact swamps the recording;
- the tissue response to chronic stimulation is not the same as the response to chronic recording;
- the user must consent to a system that can *act inside* the nervous system, not only listen.

Chapter 4 will place both directions into the same loop diagram. Chapter 10 will treat rehabilitation as one justified use of that loop. Chapter 12 will treat the consent standard as higher than for a listen-only cap.

## 3.6 Matching modality to user: worked identifications

**Case A — Amina, communication at home, no surgery accepted.**  
Research or carefully set-up EEG; reactive P300 or SSVEP if vision is reliable; perhaps a hybrid with a residual muscle switch. Invasiveness is not “better” if she will not undergo it. Bit-rate is secondary to a system a caregiver can mount.

**Case B — Amina, later, enrolled in an attempted-speech trial.**  
Intracortical or surface arrays over speech-related cortex. Information potential rises. So do obligation: device maintenance, explantation policy, what happens when funding ends.

**Case C — Stroke rehabilitation in clinic.**  
EEG or limited implanted systems paired with robotics or stimulation. The modality is chosen for a loop that can be repeated, not for a headline bandwidth.

**Case D — Office “focus” headband.**  
Consumer EEG. Coarse state estimation at best. Medical claims are a regulatory matter, not a branding matter. Workplace reading of the same signal is a consent matter (Chapter 11).

If two cases share a vendor name and nothing else, they are still two designs.

## 3.7 First contact with tissue is already ethics

Chapter 12 is the ethics chapter. It is not the beginning of ethical content.

- A penetrating electrode is an injury justified, if at all, by prospect of benefit and by support that outlasts a paper.
- A trial that explants because the grant ended has used a person as a platform.
- Neural data are not ordinary telemetry. Even coarse EEG can support inferences a user did not intend to offer.
- “Minimally invasive” is a surgical comparison, not a moral anaesthetic.

> **Definition 3.4 (Abandoned hardware).**  
> An implanted or highly customised system for which funding, maintenance, or clinical responsibility has ended while the person still lives with the consequences of having received it.

Design that cannot name an owner for year five has not finished the resolution–risk analysis.

## 3.8 What to carry into Chapter 4

Chapter 4 assembles sensors and stimulators into a loop. Carry four sentences.

1. Choose a point on the resolution–risk surface for a named user; do not choose a modality because it is fashionable.
2. EEG measures a blurred field; arrays can measure spikes; neither measures a proposition.
3. Stimulation is a different physics from recording and a different consent.
4. The front-end can destroy information that no later model will recover.

Amina’s Tuesday depends more on a stable contact and a responsible owner than on the name of a neural network.

## 3.9 Chapter summary

Recording modalities differ in the physical quantity they sample, in spatial and temporal grain, and in the burden they impose. Stimulation writes current or other energy into tissue; it is not decoding run backwards. Bidirectional systems raise artefact and consent together. The correct modality is the one that changes a named user’s options at a burden that user and institution can carry for the life of the device, not the one that maximises a laboratory figure of merit.
