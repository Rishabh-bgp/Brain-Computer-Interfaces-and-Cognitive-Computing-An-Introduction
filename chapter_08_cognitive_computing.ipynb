## 8.1 Three objects, written again with stages

> **Definition 8.1 (Cognitive computing).**  
> Design of computational systems that estimate, emulate, or adapt to cognitive functions — attention, working memory, decision, learning — whether or not a living brain is in the loop. In this book it appears at **S6** (priors, grammars, interface policies) and **S9** (when to recalibrate, when to stop).

> **Definition 8.2 (Neuromorphic computing).**  
> Hardware and algorithms that imitate neural signalling or architecture (spikes, colocated memory and compute, event-driven circuits). A possible implementation of **S3–S5** near the sensor. Not a type of BCI.

> **Definition 8.3 (Cognitive BCI).**  
> An interface whose *target variable* $y$ is a cognitive state or content — attention level, remembered item, decision variable, affect — rather than a motor or linguistic output. The loop is the loop of Chapter 4. The claim about $y$ is what changed.

A language model that rescores Amina’s next character is Definition 8.1 at S6. A spiking chip that computes CSP is Definition 8.2 at S4. An EEG system that claims to output the number she is holding in mind is Definition 8.3, and it must meet Definition 2.3 (usable trace) on a leakage-safe split.

Write the three sentences on the inside cover of a notebook. Most public confusion is a failure to keep them apart.

## 8.2 What theories may be used for

Computational cognitive science offers models that are more structured than a generic classifier.

**Attention as selection.** Limited capacity; competition among sources; load that rises with clutter and with time on task. Use: slow the P300 matrix when inferred load is high; do not flash twelve targets when six will do.

**Working memory as a limited buffer.** Capacity on the order of a few items; interference; decay. Use: design the interface so Amina need not hold a long prefix in mind if the language model already does. Do not claim that a scalp band ratio *is* the buffer.

**Decision as accumulation.** Evidence for options integrates to a threshold. Use: decide how many P300 repetitions to collect before committing a character — an explicit accumulator at S6. The cell below is that idea in one dimension.

**Learning as co-adaptation.** Already named in Definition 4.5. Use: schedule training; expect strategy.

These theories earn their keep when they change a line on the design sheet of Section 4.11. They do not earn their keep as captions on a scalp topography.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(8)
plt.rcParams.update({"figure.dpi": 110, "font.size": 11, "axes.grid": True, "grid.alpha": 0.3})


def accumulate(evidence, threshold=3.0):
    '''Drift-diffusion cartoon: stop when abs(sum) hits threshold.'''
    s = np.cumsum(evidence)
    hit = np.where(np.abs(s) >= threshold)[0]
    t_hit = int(hit[0]) if hit.size else len(s) - 1
    return s, t_hit


# Each flash contributes a noisy log-likelihood increment for 'this is the target'
ev_target = 0.35 + 0.8 * np.random.randn(25)
ev_other = -0.10 + 0.8 * np.random.randn(25)
s_t, ht = accumulate(ev_target)
s_o, ho = accumulate(ev_other)

fig, ax = plt.subplots(figsize=(7.4, 3.2))
ax.plot(s_t, label="attended flashes (should hit +threshold)")
ax.plot(s_o, label="other flashes")
ax.axhline(3.0, color="black", linestyle="--", linewidth=1)
ax.axhline(-3.0, color="black", linestyle="--", linewidth=1)
ax.scatter([ht], [s_t[ht]], color="sienna", zorder=3)
ax.set_xlabel("flash index (repetitions cost time)")
ax.set_ylabel("accumulated evidence")
ax.set_title("S6 as an accumulator: stop collecting P300 flashes when ready")
ax.legend()
plt.tight_layout()
plt.show()
print(f"target sequence committed after {ht+1} flashes; other after {ho+1}")
```

The figure is not a brain. It is a *policy*: Stage S6 stops spending Amina’s seconds when the posterior is good enough. That is cognitive computing in the only sense this book needs.

## 8.3 Why cognitive $y$ is harder than motor or oddball $y$

Motor imagery and P300 work, when they work, because the experimenter defines a generating mechanism that is relatively local and repeatable: a sensorimotor rhythm, a time-locked oddball. Cognitive variables are different.

- **Distributed.** Working memory and value are not known to sit as a stable rate in one channel (Chapter 2).  
- **Mixed.** The same population participates in many tasks.  
- **Contextual.** “Attention” in a dark lab is not “attention” in a ward.  
- **Unlabelled in life.** There is no cue that tells you what Amina is holding in mind unless you supplied the item.  
- **Easy to fake with physiology.** Fatigue, time of day, and EMG correlate with questionnaires.

A decoder that classifies high versus low load on a laboratory $n$-back has found a usable trace of a *coarse state*, if the split is honest. It has not recovered contents. Contents — *which* digit, *which* word — are a different $y$ and almost never available from scalp EEG.

> **Definition 8.4 (State versus content).**  
> A *state* variable is a coarse coordinate such as load, vigilance, or valence. A *content* variable is the item being processed. Cognitive BCI claims must say which one they mean.

Most wearable products advertise content and deliver, at best, a noisy state.

## 8.4 Passive BCI as policy, not as telepathy

Passive systems estimate a state the user did not encode as a command (Definition 1.4). Used well, they change S6:

- pause the lesson when vigilance collapses;  
- reduce flicker when inferred visual fatigue rises;  
- refuse to issue a new character while an error-related potential suggests Amina has noticed a mistake.

Used badly, they emit a score into a workplace (Chapter 11). The engineering may be identical. The loop is not: a supervisor has been inserted at S8.

Cognitive computing here is the policy that maps $\hat{y}_{\text{state}}$ to an interface action, including the action *do nothing and do not log this*.

## 8.5 Neuromodulation as the other half of cognition

Decoding is not the only path to cognitive $y$. Clinical neuromodulation already changes mood, tremor, and seizure likelihood by writing to tissue (Chapter 3). Closed-loop variants measure a state and stimulate.

This path has two lessons for cognitive BCI.

1. **Causal leverage.** If stimulating a circuit changes a cognitive variable, that circuit is more than a correlate. That is stronger evidence than an EEG classification of a questionnaire.  
2. **Different ethics.** A system that writes can harm in ways a listen-only cap cannot. Consent, abort criteria, and ownership of the device (Definition 3.4) are part of the design sheet.

An introductory course will not teach how to build a psychiatric closed loop. It will teach that “cognitive BCI” includes this family, and that decoding-only posters are not the whole subject.


---

## 8.6 Neuromorphic hardware: implementation, not metaphor

Spiking chips, in-memory compute, and event-driven sensors can lower the joules and the watts of S3–S5. For implanted systems, power is a clinical constraint. For wearables, battery life is a usability constraint.

None of that implies that the decoder “thinks,” that spikes on a chip are spikes in Amina, or that organoid devices (if mentioned at all) belong in Chapter 1’s definition. Put neuromorphic claims in the hardware column of the design sheet. If they do not change latency, power, or risk, they are decoration.

## 8.7 Worked placements

**Language model on Amina’s speller.** Definition 8.1, S6 prior $P(\text{next character} \mid \text{prefix})$. Must be ablated (Chapters 5 and 7).

**Accumulator that stops P300 repetitions.** Definition 8.1, S6 policy. The cell above.

**Workload scalar that slows a training game.** Passive BCI plus Definition 8.1 policy. State, not content (Definition 8.4).

**Attempted-speech decoder.** Not a cognitive BCI in the sense of 8.3 if $y$ is words she is trying to say; that is a linguistic-motor output. The language model on top is 8.1.

**EEG “mind-reading of inner speech” product.** Almost always an 8.3 *content* claim without a usable trace. Refuse it until Definition 2.3 is met on held-out sessions.

**Closed-loop stimulation triggered by a decoded high-risk state.** Bidirectional loop; 8.5; ethics of writing.

## 8.10 Chapter summary

Cognitive computing, neuromorphic computing, and cognitive BCI are neighbours. The first supplies priors and policies at S6 and S9. The second supplies a hardware option at S3–S5. The third changes the target variable $y$ to a cognitive state or — much harder — to content. Cognitive $y$ is distributed, mixed, and poorly labelled. Closed-loop stimulation is a second, ethically heavier path. Theories of attention, memory, and decision are welcome when they change the design sheet. They are not welcome as synonyms for a scalp feature.
