## 1.1 A scene, then a definition

Consider a person who can no longer speak or type. An electrode array, or a cap of scalp sensors, records electrical activity produced by populations of cortical neurons. A computer transforms that activity, trial by trial or continuously, into a letter, a word, a cursor movement, or a synthesised utterance. Nothing in the person’s peripheral motor system is required for the command to appear. The pathway from intention to artefact has been rerouted through a machine.

That scene contains the essential pieces of a **brain–computer interface (BCI)**:

1. a measurement of central-nervous-system activity;
2. a computational transformation of that measurement;
3. an output that does useful work in the world or in a therapy;
4. typically, a user who can perceive the output and therefore close a loop.

A definition used throughout this book, adapted from the consensus language of the field, is the following.

> **Definition 1.1 (Brain–computer interface).**  
> A brain–computer interface is a system that measures activity of the central nervous system and converts it into an artificial output that replaces, restores, enhances, supplements, or improves natural CNS output, or that serves as a research tool for studying the nervous system.

Four remarks prevent the definition from collapsing into marketing.

**Remark A — The measurement must be of the CNS.**  
Electrical activity of muscles (electromyography), movement of the eyes, or respiration can be excellent control signals. They are not BCI signals. An interface that secretly relies on forehead EMG while advertising “thought control” is not a BCI, however convenient the confusion may be for a demonstration.

**Remark B — The output is artificial.**  
Speech, reaching, and walking are natural CNS outputs. A spelling matrix, a robotic arm, a stimulator pulse, or a change in a user-interface layout is an artificial output. The definition is silent on whether that output is impressive. A system that turns a reliable brain signal into a single binary alarm satisfies the definition.

**Remark C — The user is usually in the loop.**  
Most practical BCIs are closed loops: the user sees or hears the result and adjusts. Offline decoding of archived recordings is BCI *research*; it is not yet an interface. Chapter 4 develops this distinction.

**Remark D — “Thought” is not a technical term in this book.**  
What is decoded is a statistical pattern associated with a task the experimenter has defined: imagined movement of the right hand, attention to a flashing letter, attempted articulation of a word. That pattern may be related to a conscious intention. The decoder does not inspect a proposition in a language of thought. Section 1.2 returns to this point.


### Adjacent terms

| Term | Use in this book |
|------|------------------|
| **BCI** | Default term for the interface defined above. |
| **BMI** (brain–machine interface) | Used interchangeably in much of the literature; here reserved, when a distinction is useful, for systems whose output is a physical actuator rather than a symbolic interface. |
| **Neural interface** | Broader class that includes BCI, nerve interfaces, and many stimulating implants. |
| **Neuroprosthesis** | Device that substitutes a lost sensory or motor function; a cochlear implant is a neuroprosthesis that is not ordinarily called a BCI. |
| **Cognitive computing** | Engineering and modelling programme that aims at systems with attention, memory, or decision-like behaviour. Related to BCI; not identical with it. See Section 1.6. |
| **Neuromorphic computing** | Hardware and algorithms that imitate neural signalling and architecture. An implementation option, not a type of BCI. |

## 1.2 What a BCI is not

Three neighbouring technologies are routinely described as if they were BCIs. Drawing the boundary is not pedantry. It determines which experimental controls, which risk disclosures, and which performance metrics are appropriate.

**Eye-tracking and residual movement.**  
Many people who cannot speak can still move the eyes. A camera that converts gaze into a cursor can restore communication at high bit-rates. It is an assistive interface. It is not a BCI, because the measured variable is extra-ocular muscle and gaze geometry, not CNS activity recorded as such. A fair comparison of a P300 speller with a gaze speller must admit that the latter often wins on speed; the BCI is justified when gaze is unavailable or unstable.

**Electromyography and mechanical sensors.**  
Forehead, jaw, and neck muscles leak into EEG. A system that classifies “clench” versus “relax” from a dry forehead band may be useful. Calling it a BCI without artefact analysis is a category error. Chapter 6 treats artefact as a first-class object precisely because this error is common.

**Imagined total access to mental content.**  
Popular descriptions speak of uploading memories, of a transcript of inner speech, or of a device that knows what a person “really thinks.” No current recording modality supports those claims, for reasons already stated in Chapter 0: electrodes observe mixtures, codes are distributed, and the mapping from content to field is many-to-one. A responsible introductory text refuses that vocabulary even when a decoded sentence is impressive.

## 1.3 A short historical ladder

A complete history is a book of its own. The following six rungs are sufficient to see that BCI is not a sudden invention of the 2020s and that each rung added a different engineering problem.

**Hans Berger and the electroencephalogram (1920s).**  
Berger showed that electrical activity of the human brain can be recorded from the scalp. Without that measurement there is no non-invasive BCI. The scientific object was clinical and physiological, not an interface.

**Jacques Vidal and the naming of the problem (1970s).**  
Vidal’s work at UCLA framed “direct brain–computer communication” as an engineering programme: treat the EEG as an input channel and close a loop with a computer. The conceptual move matters more than any particular accuracy he obtained. The field acquired a name and a pipeline.

**Event-related potentials and the P300 speller (1980s–1990s).**  
Farwell and Donchin, among others, showed that a time-locked scalp response to a rare, attended stimulus can select items from a matrix. Communication without movement became a laboratory fact. The new problems were rate, fatigue, and reliability outside a shielded room.

**Operant conditioning of rhythms and motor imagery.**  
Work associated with Wolpaw and others showed that people can learn to modulate sensorimotor rhythms and that those rhythms can move a cursor. The user is not a static source. Training and co-adaptation entered the centre of the subject.

**Intracortical arrays in animals and then in humans.**  
Microelectrode arrays in motor cortex enabled decoding of intended kinematics with a bandwidth that scalp EEG does not provide. Translation to people with paralysis turned BCI from a signal-processing curiosity into a clinical research enterprise, with surgical risk and long-term tissue response as new constraints.

**Speech, endovascular sensors, and consumer devices (contemporary horizon).**  
High-density cortical recordings plus language models have produced attempted-speech systems whose output resembles conversation more than a matrix speller. Minimally invasive vascular sensors and a parallel industry of wearable EEG have split the public story into two speeds: high-bandwidth clinical systems and low-bandwidth consumer systems that share a name. Chapter 11 insists on keeping those speeds apart.

This ladder is not an argument that later rungs make earlier ones obsolete. A well-built P300 speller remains the right answer for some users. Novelty is not a clinical virtue.

## 1.4 Three taxonomies

A single label such as “an implant” or “an EEG headset” under-describes a system. This book classifies every example on three independent axes.

### 1.4.1 Invasiveness

| Class | Where the sensor sits | Typical cost in risk and logistics |
|-------|----------------------|------------------------------------|
| Non-invasive | Skin (EEG, fNIRS) or outside the head (MEG) | Setup, artefacts, limited spatial grain |
| Semi-invasive | Inner skull surface or vascular tree near cortex (ECoG, some stent-mounted sensors) | Surgery of varying gravity; better signal than scalp |
| Invasive | Penetrating cortical or deeper electrodes | Highest information potential; surgical and chronic-tissue cost |

Invasiveness is not a moral ranking and not a performance ranking by itself. It is a trade-off surface. Students should be able to defend a choice on that surface for a named user, not to recite that “invasive is always better.”

### 1.4.2 Direction of information

- **Brain-to-computer (the classical BCI):** neural activity is decoded into a command or a state estimate.
- **Computer-to-brain:** patterned stimulation is delivered to alter neural activity (many neuromodulation devices live here; not all are called BCIs).
- **Bidirectional:** decode and stimulate in one loop. Closed-loop stimulation for seizures or for motor rehabilitation belongs here. So does any future system that writes sensory feedback into cortex while reading motor intent.

A device that only stimulates on a fixed schedule is not thereby a BCI. The interface character appears when measured neural activity governs the stimulation, or when stimulation is used to complete an interactive loop with a decoder.

### 1.4.3 Purpose (six themes)

A widely used map of applications, sometimes called the BNCI Horizon themes, is sufficient for an introductory course:

1. **Replace** — substitute a lost natural output (for example a communication channel).
2. **Restore** — return a function toward its previous form (for example grasp via neural control of stimulation).
3. **Enhance** — improve a function already present (contested; treat claims sceptically).
4. **Supplement** — add a channel the nervous system did not have (for example a silent alarm).
5. **Improve** — use the closed loop as therapy so that natural function itself changes (rehabilitation).
6. **Research tool** — use the loop to study the nervous system rather than to serve the user.

A single hardware platform can serve more than one theme. The theme changes the ethical and evaluative standard. An enhancement claim in a healthy consumer requires a different burden of proof from a replacement claim in locked-in syndrome.


```python
import numpy as np
import matplotlib.pyplot as plt

themes = [
    "Replace",
    "Restore",
    "Improve",
    "Supplement",
    "Enhance",
    "Research",
]
# Illustrative only: relative emphasis in an introductory clinical-first course
weights = np.array([5, 5, 4, 3, 2, 4], dtype=float)
weights = weights / weights.sum()

fig, ax = plt.subplots(figsize=(7.2, 3.4))
ax.bar(themes, weights, color="#4c6a92")
ax.set_ylabel("emphasis in this book (relative)")
ax.set_title("Application themes — pedagogical emphasis, not market share")
ax.set_ylim(0, 0.3)
plt.tight_layout()
plt.show()
```

## 1.5 Active, reactive, and passive interfaces

A second, orthogonal classification concerns how the brain activity is elicited.

> **Definition 1.2 (Active BCI).**  
> The user voluntarily produces a covert action that the decoder is trained to recognise — for example motor imagery, or attempted speech.

> **Definition 1.3 (Reactive BCI).**  
> The system presents stimuli; the decoder uses the brain’s evoked response to those stimuli. The P300 speller and SSVEP selectors are reactive.

> **Definition 1.4 (Passive BCI).**  
> The system estimates a state the user is not trying to encode — workload, fatigue, affect — and adapts the environment. No command is issued by the user *as a command*.

These three are often mixed. A communication system may be reactive for letter selection and passive for pausing when inferred fatigue is high. Cognitive computing, in the sense of this book, is most at home in the passive and adaptive layer: it supplies models of state and policies for changing the interface. It is not required for an active motor-imagery classifier to exist.

**Worked identification.**  
Amina, who will recur in later chapters, has amyotrophic lateral sclerosis. On Monday she uses a P300 matrix to spell (reactive, replace, non-invasive). On Thursday a clinical team records attempted speech from cortex (active, replace, invasive). In both cases a language model can sit downstream as a prior (Chapter 0, Bayes). The language model is cognitive computing in a modest, defensible sense: it is a model of linguistic expectation, not a model of Amina’s mind.

## 1.6 Users and settings

Name the user before you name the algorithm.

**Clinical users with severe motor or speech impairment.**  
Here the comparison class is a letter board, a gaze tracker, or silence. Reliability, setup time, and caregiver burden dominate bit-rate. Abandonment after a research trial is an ethical failure mode, not a footnote (Chapter 12).

**Rehabilitation users.**  
The aim is not a permanent substitute channel but a change in the nervous system or in residual function. Evaluation must follow clinical trial logic, including control conditions that isolate placebo and therapist attention.

**Research participants.**  
Students and laboratories often test on healthy adults. That is legitimate for method development. It is not evidence that a system will serve Amina.

**Consumer and workplace users.**  
Attention headbands, games, and proposed workplace monitors live here. Consent, function creep, and the difference between a medical claim and a wellness claim are the governing issues. Chapter 11 is devoted to them.

A design that is optimal for a seated laboratory participant with a technician holding the cap is not thereby a design for any of the groups above.

## 1.7 Cognitive computing as a neighbouring field

The phrase *cognitive computing* has been used, at various times, for rule-based cognitive architectures, for IBM’s commercial branding, for neuromorphic hardware, and for almost any adaptive software. This book uses it narrowly.

> **Definition 1.5 (Cognitive computing, working sense).**  
> The design of computational systems that estimate, emulate, or adapt to cognitive functions — including attention, working memory, decision, and learning — whether or not those systems interface with a living brain.

Three relationships to BCI follow.

1. **Theories as sources of features and policies.**  
   An accumulator model of decision, or a limited-capacity model of attention, can suggest what to decode and when to change trial pace. That is theory serving the loop of Chapter 4.

2. **Passive and neuroadaptive interfaces.**  
   If the interface estimates workload and simplifies a display, it is a BCI wrapped in a cognitive-computing policy.

3. **Neuromorphic implementation.**  
   Spiking chips and in-memory compute may eventually sit next to an electrode as a low-power front-end. They do not make the decoder a brain.

> **Warning.**  
> Do not write “the BCI is a cognitive computer” as if the sentence had content. A BCI is an interface. A cognitive computer is a machine or a model. Their intersection is a research programme, which Chapter 8 treats in detail.

Cognitive *BCI* is a still narrower phrase: an interface whose target variable is itself a cognitive state (attention, memory, value), rather than a motor or linguistic output. Distributed coding and context dependence make that target harder than cursor control. The difficulty is a scientific fact, not a reason to avoid the topic.

## 1.9 A compact map of claims you should now refuse

| Claim you may meet | Precise objection |
|--------------------|-------------------|
| “The headset reads your thoughts.” | It records a mixed field and decodes a task-defined pattern. |
| “Non-invasive equals safe and equivalent.” | Risk is lower; information grain is not equivalent. |
| “Invasive equals clinically ready.” | Surgical feasibility is not home use, maintenance, or consent over years. |
| “Accuracy of 95 percent.” | On which split, which baseline, which user, which session? |
| “AI has solved BCI.” | Models improved decoding; they did not remove calibration, drift, or ethics. |
| “Cognitive computing *is* BCI.” | One is a modelling and systems programme; the other is an interface. |

## 1.10 Chapter summary

A BCI measures central-nervous-system activity and turns it into an artificial output in a loop that a person can, in the typical case, perceive. Adjacent tools — gaze, EMG, classical sensory implants — may be more useful in a given life and still not be BCIs. History added measurement, a named pipeline, evoked and imagery paradigms, implants, and speech-oriented systems; it did not replace the need to name the user. Classify every example by invasiveness, direction, and purpose, and by whether it is active, reactive, or passive. Treat cognitive computing as a neighbour that supplies models and policies. Refuse the sentence “the machine read a thought” until you can replace it with a sentence about a signal, a task, and a decoder.
