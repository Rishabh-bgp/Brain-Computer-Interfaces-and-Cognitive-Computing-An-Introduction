## 12.1 Evaluation as a bundle, not a percentage

> **Definition 12.1 (Reporting bundle).**  
> The minimum set of facts that makes a BCI result interpretable: user, purpose, paradigm, modality, stages S0–S9, split, chance, baseline, artefact audit, calibration burden, credit split, user-facing metric, harms, and owner after the study.

If an item is missing, the number that remains is a demonstration.

**Checklist against this book**

| Item | Where defined |
|---|---|
| User and setting | Ch. 1, 9, 11 |
| Purpose theme; active/reactive/passive | Ch. 1 |
| Point on resolution–risk surface | Ch. 3 |
| Paradigm family | Ch. 5 |
| Causal $\Phi$; nested parameters | Ch. 6 |
| Split and subject-independence claim | Ch. 7 |
| Chance, linear baseline, balanced error | Ch. 7 |
| Loop latency budget | Ch. 4 |
| ITR only as a side comparison | Ch. 9 |
| User-facing metric; day table | Ch. 9 |
| Credit split S5 / S6 / helper | Ch. 9 |
| Sham or matched dose if improvement claimed | Ch. 10 |
| Wearer- vs other-directed | Ch. 11 |
| Owner in year five if implanted | Ch. 3 |

A conference paper need not contain every row at monograph length. It must not *contradict* a row by silence when the claim requires that row.


```python
def bundle_score(present):
    '''present: dict of item -> bool. Returns fraction complete and missing names.'''
    missing = [k for k, v in present.items() if not v]
    frac = 1.0 - len(missing) / max(len(present), 1)
    return frac, missing


example = {
    "user": True,
    "purpose": True,
    "split": False,
    "chance": True,
    "baseline": False,
    "artefact_audit": False,
    "calibration_min": False,
    "credit_split": False,
    "user_metric": False,
    "owner_year5": False,
}
frac, missing = bundle_score(example)
print(f"bundle completeness: {frac:.0%}")
print("missing:", ", ".join(missing))
print("A 94 percent accuracy attached to this bundle is not a result.")
```

## 12.2 Translation is a path with gates

Laboratories produce recordings. Homes produce Tuesdays. Between them sit gates that differ by country and by claim class (Definitions 11.3 and 11.4). The book uses a generic ladder, not a statute book.

1. **Preclinical / healthy-user methods.** Fine for $\Phi$ and $f_\theta$. Not evidence for Amina.  
2. **Feasibility in the target users.** Safety, whether the generating mechanism exists, whether the day table is livable.  
3. **Pivotal evidence.** Pre-registered endpoints; comparators; enough people to bound harm.  
4. **Regulatory authorisation** for a *stated intended use*. Authorisation is not a moral halo and not a licence to advertise a different $y$.  
5. **Reimbursement and service.** Who pays for the cap, the surgery, the technician, the replacement battery.  
6. **Maintenance.** Firmware, explantation, a telephone number that answers.

> **Definition 12.2 (Intended use).**  
> The sentence that names the user, the purpose theme, and the output. Evaluation and ethics are relative to this sentence. Changing the sentence after authorisation is a new device.

Students who write “we will get FDA/CE/CDSCO approval” as a milestone without naming intended use have written a wish.

## 12.3 Consent is a process with a power structure

Consent is not a signature. It is a process that can fail while the form succeeds.

**Capacity and dependence.** Amina may depend on the team for hope, transport, or care. That is not peer review.

**What must be sayable in ordinary words.** Surgery and explant; what $\hat{y}$ will be used for; who else will see it; how to leave; what happens if the grant ends; that inner speech is not promised.

**Ongoing refusal.** A loop that cannot be switched off by the wearer is a different object (Definition 11.2).

**Other-directed systems.** Workplace consent under hierarchy is weaker than the word “consent” suggests (Chapter 11). Do not launder it through a student ethics form.

Cognitive BCI and writing to tissue raise the bar further: you are asking to touch the organ of the agreement itself.

## 12.4 Neural data and mental privacy

> **Definition 12.3 (Neural data).**  
> Recorded activity of the nervous system, plus any features, models, and labels derived from it that can be linked to a person.

Neural data are not ordinary telemetry.

- They can support inferences the wearer did not offer (affect, sleep, impairment).  
- They can outlive the study in a pretrained model (Chapter 7, data labour).  
- They can be re-identified when combined with date, clinic, or rare disease.

**Working rules for this course**

1. Collect the minimum $C$ and $T$ the intended use needs.  
2. Separate identity from traces; document who can join them.  
3. Freeze purposes; secondary use needs a new agreement.  
4. Do not ship raw traces to a public model card without a legal and ethical path.  
5. Deletion must be possible for identifiable recordings; “the weights already absorbed you” is not an answer you may hide.

Mental privacy, in the modest sense used here, is the right not to have $y_{\text{content}}$ inferred or advertised without a usable-trace justification and an agreement. Neurorights statutes are emerging and uneven. Design as if a sceptical court will read Definition 8.4.

## 12.5 Abandoned hardware and the owner

Definition 3.4 named abandoned hardware. Translation makes it operational.

An implant trial must name, *before first surgery*:

- who pays for revision and explant;  
- who maintains software when the start-up dissolves;  
- how long $\theta$ will be updated;  
- what the wearer may keep if they leave the study;  
- a successor owner if the principal investigator leaves.

A consumer cap that bricks itself when a subscription ends is a mild form of the same structure. Do not treat mildness as innocence if $\hat{y}$ has become part of someone’s workday.

## 12.6 Equity of access

The resolution–risk surface is also a wealth surface. Neurosurgery, MEG, and high-bandwidth speech trials cluster in a few hospitals. Consumer EEG clusters in shops. Languages and scripts that S6 does not know are excluded twice (Chapter 9).

An introductory textbook cannot fix allocation. It can forbid a sentence that calls a system “universal” when the day table requires a technician and an English prior.

> **Definition 12.4 (Access-complete design).**  
> A design sheet that names who cannot use the system: by language, by cost, by surgical eligibility, by vision, by caregiver time.

If that list is empty, the sheet is false.

## 12.7 Dual use and enhancement, restated

Recording and stimulation can serve communication and they can serve interrogation fantasies, battlefield monitoring, or coercive evaluation (Definition 11.5). Students will be asked to work on “human performance.”

The book’s test is ordinary and strict:

- Is the loop wearer-directed?  
- Is $y$ a usable trace or a slogan?  
- Can the wearer refuse?  
- Would you accept the system if $\hat{y}$ were about you and visible to someone who grades you?

If the answers fail, the project is not saved by calling it cognitive computing.

## 12.8 A research agenda sized for a master’s year

Pick one. Finish the bundle.

1. Leakage-safe comparison of regularised LDA versus a small network on public motor-imagery data, with an artefact audit.  
2. Causal-filter P300 pipeline with a measured latency budget and an undo cell.  
3. Credit-split reanalysis of a speech or wheelchair paper using only published numbers plus stated assumptions.  
4. Language-prior ablation on a small vocabulary, reporting S5 alone.  
5. Day-table study: setup and teardown timed on healthy users wearing a research cap; no clinical claim.  
6. Sham-gating protocol *written*, not run, for a local rehab partner.  
7. Workplace-policy critique of one consumer SDK against Definitions 11.2 and 12.3.

These are not dissertations. They are how a first-semester student joins the field without adding a poster that this chapter would reject.

## 12.9 What to carry into the laboratory appendix

Appendix L will ask you to touch public data. Carry the bundle, not the mood.

1. Name the claim before the score.  
2. Intended use is a sentence.  
3. Consent and data rules apply to public datasets too (licences, no re-identification games).  
4. No implant work in the appendix.  
5. A missing owner is a missing design.

## 12.10 Chapter summary

Evaluation is a bundle of facts that makes a number mean something on Tuesday. Translation is a gated path relative to an intended use, not a synonym for a press release. Consent is a process under power. Neural data can outlive the session inside other people’s models. Abandoned hardware is a foreseeable harm. Access is part of the design sheet. Enhancement and other-directed evaluation are not saved by technical novelty. The work that remains for a student is small, named, and leakage-safe.
