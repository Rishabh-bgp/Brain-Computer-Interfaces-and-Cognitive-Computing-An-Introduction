# Chapter 10 — Rehabilitation and Closed Loops

**Book:** *Brain–Computer Interfaces and Cognitive Computing: An Introduction*  
**Prerequisites:** Chapters 0–9.  
**Author:** Er. Rishabh Aryan

Chapter 9 used the loop to **replace** an output. This chapter uses it to **improve** residual function, or to claim that it does. The actuator may look the same — a robot, a stimulator, a cursor. The success criterion does not. Amina-as-communicator is finished when a message arrives. A person after stroke is finished only if *their own* reaching changes after the machine is unplugged, or if a pre-registered functional score says so.

**Learning outcomes**

1. Distinguish replacement, restoration, and improvement as purpose themes with different endpoints.
2. Describe pairing of decoded intent with robotics or stimulation as a plasticity hypothesis, not as a demonstration of force.
3. Sketch a trial that can isolate BCI from therapist attention and from placebo.
4. Treat neurofeedback and psychiatric targets as cousins with a higher hype-to-evidence ratio.
5. Refuse a rehabilitation claim that reports only online classification accuracy.


---

## 10.1 Three endings of the loop

> **Definition 10.1 (Replacement endpoint).**  
> The artificial output works while the system is on. Chapter 9.

> **Definition 10.2 (Restoration endpoint).**  
> The artificial output substitutes a function in daily life — FES grasp, a worn robot — still dependent on the device.

> **Definition 10.3 (Improvement endpoint).**  
> After training, a clinical or kinematic measure of *the person’s own* function changes by a pre-registered amount, with controls that bound therapist and placebo effects.

A paper that trains motor imagery for four weeks and quotes 85 percent decoding accuracy has not addressed Definition 10.3. It has addressed whether S5 still works.

Cognitive computing sits here as a **dosing policy**: when to assist, when to fade assistance, when to stop a session because inferred fatigue makes plasticity claims implausible. It does not sit here as a theory that “the model rewired the cortex.”


---

## 10.2 The pairing hypothesis

Rehabilitation BCI often rests on a simple physiological bet.

If a decoded motor intention is delivered, promptly, as a proprioceptive or electrical consequence — robot-assisted movement, functional electrical stimulation of the same limb — then Hebbian and use-dependent plasticity may strengthen residual pathways.

The bet needs three engineering facts from earlier chapters:

- a generating mechanism the user can still produce (usually sensorimotor activity, Chapter 5);  
- a loop latency short enough that the consequence still belongs to the intention (Chapter 4);  
- a sensor the clinic can mount every session (Chapter 3).

> **Definition 10.4 (Pairing).**  
> A closed-loop mapping from a decoded motor-related state to a somatosensory or motor consequence on a timescale relevant to synaptic change.

If the robot moves on a fixed schedule regardless of $\hat{y}$, you have robotics, not pairing. If $\hat{y}$ is forehead EMG, you have a muscle-triggered exerciser. Both may help patients. Neither is evidence for the BCI hypothesis.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)
plt.rcParams.update({"figure.dpi": 110, "font.size": 11, "axes.grid": True, "grid.alpha": 0.3})

# Cartoon: probability of a 'good' pairing falls as loop delay grows
delay_ms = np.linspace(0, 800, 17)
# illustrative, not a measured curve
p_credit = np.exp(-delay_ms / 280.0)

fig, ax = plt.subplots(figsize=(7.2, 3.2))
ax.plot(delay_ms, p_credit, color="#4c6a92", marker="o")
ax.set_xlabel("loop delay from intent feature to consequence (ms)")
ax.set_ylabel("illustrative pairing credit")
ax.set_title("If the hand moves too late, the cortex may not bind the events")
plt.tight_layout()
plt.show()
print("Fill this curve with data before you claim a window; the shape is the design constraint.")
```

## 10.3 Partners in the clinic

**Robot-assisted therapy.** High repetition, measurable kinematics, fatigue of therapists reduced. The BCI’s job is to gate assistance on intention, and later to *fade* assistance so the person does the work.

**Functional electrical stimulation (FES).** Muscles contract in a grasp pattern. Pairing decoded intention with FES is a restoration path (10.2) that may also aim at improvement (10.3).

**Therapists.** They are in the loop. Their attention is an intervention. Any trial that adds BCI plus extra enthusiastic hours and then compares to “usual care” has confounded S8 with a social treatment.

**Medication, spontaneous recovery, and time.** Subacute stroke recovers partly on its own. Uncontrolled before–after series will attribute the calendar to the cap.


---

## 10.4 Who the user is

Not Amina-as-communicator, or not only.

- **Stroke**, especially hemiparesis with residual cortical signal.  
- **Spinal cord injury**, where pairing may target residual pathways or operate FES.  
- **Other motor disorders** case by case.

Exclusion and inclusion must name whether the generating mechanism of Chapter 5 is still available. A completely isolated cortex with no sensorimotor modulation cannot enter an imagery-paired protocol except as research on a different question.

Fatigue, depression, aphasia, and neglect change S8. They are not noise. They are the user.


---

## 10.5 Evidence that deserves the word

A rehabilitation claim that can survive this book looks roughly like the following.

1. Pre-registered primary endpoint (for example a Fugl–Meyer change, a timed reach, a specified functional task).  
2. Comparator: robotics or FES without neural gating, or sham neural gating, plus matched therapist time.  
3. Allocation described; assessor blinded where possible.  
4. BCI-specific process measures *secondary*: accuracy, pairing delay, dose of successful pairs.  
5. Follow-up after the device stops.  
6. Harm: fatigue, frustration, skin, seizure risk if stimulation is used.

> **Definition 10.5 (Sham gating).**  
> A control in which the user and the robot behave as in the BCI arm but $\hat{y}$ does not determine the consequence, or determines it at a delay or a randomness that breaks pairing.

Without some version of 10.5, you cannot separate the pairing hypothesis from wearing a cap in a motivated room.


```python
# Tiny illustration: why before-after is not a trial
n = 24
rng = np.random.default_rng(10)
# everyone drifts up a little (spontaneous / calendar)
calendar = rng.normal(2.0, 1.0, n)
bci_extra = rng.normal(1.2, 1.5, n // 2)
ctrl_extra = rng.normal(1.0, 1.5, n // 2)  # matched therapy
delta_bci = calendar[: n // 2] + bci_extra
delta_ctrl = calendar[n // 2 :] + ctrl_extra
print("mean change BCI arm   ", round(float(delta_bci.mean()), 2))
print("mean change control   ", round(float(delta_ctrl.mean()), 2))
print("mean change if you only published the BCI arm:", round(float(delta_bci.mean()), 2))
print("The last number is the one an uncontrolled series reports as 'the BCI effect'.")
```

## 10.6 Neurofeedback

Neurofeedback shows a person a transform of their own signal and asks them to change it. It is a loop. It is often sold as therapy for attention, mood, or pain.

Place it correctly:

- as a research tool, it can teach controllability of a feature;  
- as a clinical product, it has a long history of claims that outrun sham-controlled evidence;  
- as a component of rehabilitation BCI, it is the feedback display at S7–S8, not a separate magic.

Demand sham signals and pre-registered endpoints here even more loudly than in pairing studies. Cognitive-computing language (“the system trains attention”) is not an endpoint.


---

## 10.7 Psychiatric and cognitive targets

Mood, obsession, addiction circuits, memory prostheses: these appear in reviews and in closed-loop neuromodulation programmes (Chapter 8). An introductory BCI course should:

- name them as **emerging, high-stakes, ethically heavy**;  
- refuse consumer EEG as a treatment of depression on present evidence;  
- note that writing to tissue (DBS-like) is closer to existing clinical practice than decoding inner affect from a headband;  
- send detailed protocols to a later, supervised course.

Placebo, rating-scale inflation, and desperation of users are not footnotes. They are the measurement model.


---

## 10.8 Dose, fade, and the policy at S6

If pairing works, dose matters: number of successful pairs per session, number of sessions, delay distribution.

Assistance should usually **fade**. A robot that always finishes the reach trains the robot. A policy that reduces assistance as residual movement appears is cognitive computing in the allowed sense (Definition 8.1): a control law over therapy, not a mind.

Log the policy. If fade is manual and unrecorded, you cannot attribute improvement to pairing.

## 10.9 What to carry into Chapter 11

Chapter 11 leaves the clinic. The same loop will be sold as wellness. Carry the habit of endpoints.

1. Name replacement, restoration, or improvement.  
2. Pairing needs delay, a real generator, and a sham.  
3. Therapist time is an intervention.  
4. Accuracy is a process measure.  
5. Psychiatric consumer claims are not Chapter 10 done at home.


---

## 10.10 Chapter summary

Rehabilitation BCI is a closed loop justified, if at all, by restoration or by improvement of the person’s own function. The central hypothesis is pairing of decoded intention with a timely motor or electrical consequence. Robots and FES are partners; therapists and calendars are confounders. Trials need pre-registered endpoints and sham or matched-dose controls. Neurofeedback and psychiatric targets share the loop diagram and not yet the evidence standard of a first-semester claim. Online accuracy is necessary bookkeeping. It is not recovery.
